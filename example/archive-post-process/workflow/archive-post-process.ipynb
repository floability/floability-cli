# Verify extracted archive data

Floability downloads and safely extracts three ZIP layouts before this notebook starts.

```python
from pathlib import Path

expected_files = {
    Path("data/nested-directories/nested-data/group-a/observations.txt"),
    Path("data/nested-directories/nested-data/group-b/deeper/measurements.txt"),
    Path("data/single-directory/single-layer/alpha.txt"),
    Path("data/single-directory/single-layer/beta.txt"),
    Path("data/single-file/single-file.txt"),
}

missing = sorted(path for path in expected_files if not path.is_file())
assert not missing, f"Missing extracted files: {missing}"

for path in sorted(expected_files):
    print(f"{path}: {path.read_text(encoding='utf-8').strip()}")

print(f"Verified {len(expected_files)} extracted files.")
```
